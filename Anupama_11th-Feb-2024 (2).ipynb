## Q 5. Find the Total Revenue generated by the E-Commerce company over all orders placed. (6 marks)

#### Hint: Calculate revenue of E-commerce company using the conditions mentioned above and then do the total (sum) of all to get total Revenue. 

### Ans 5

```python
sum = 0;
for i in data.discounted_price:
    if(i <= 100):
        sum = sum + i * 5/100;
    elif (i>100 and i<=350):
        sum= sum + i* 10/100;
    elif (i>350 and i<=600):
        sum= sum+ i* 15/100;
    elif (i>600):
        sum= sum+i* 25/100;
        
print(sum)
```

```python
data['revenue'] = 0

for i, row in data.iterrows():
    if row['discounted_price'] <= 100:
        data.at[i, 'revenue'] = row['discounted_price'] * 5 / 100
    elif 100 < row['discounted_price'] <= 350:
        data.at[i, 'revenue'] = row['discounted_price'] * 10 / 100
    elif 350 < row['discounted_price'] <= 600:
        data.at[i, 'revenue'] = row['discounted_price'] * 15 / 100
    elif row['discounted_price'] > 600:
        data.at[i, 'revenue'] = row['discounted_price'] * 25 / 100

data['brandrevenue'] = data['discounted_price'] - data['revenue']

total_brand_revenue = data.groupby('Brand')['brandrevenue'].sum()

top_10_brands = total_brand_revenue.sort_values(ascending=False).head(10)

print(top_10_brands)

```

**Q6.Calculate the total BrandRevenue and list the top 10 Brand having maximum revenue in descending order (6 marks)**
Hint: Total BrandRevenue is BrandRevenue which is generated after all the deductions. Also, BrandRevenue is different from the E-Commerce company's revenue.
Brand Revenue is the discounted price minus the Revenue (generated using the conditions given above Q5)

### Ans 6

```python
data['Revenue'] = 0
for i,row in data.iterrows():
    if row['discounted_price'] <= 100:
        data.at[i, 'Revenue'] = row['discounted_price'] * 5 / 100
    elif 100 < row['discounted_price'] <= 350:
        data.at[i, 'Revenue'] = row['discounted_price'] * 10 / 100
    elif 350 < row['discounted_price'] <= 600:
        data.at[i, 'Revenue'] = row['discounted_price'] * 15 / 100
    elif row['discounted_price'] > 600:
        data.at[i, 'Revenue'] = row['discounted_price'] * 25 / 100
        
data['brand_revenue'] = data['discounted_price'] - data['Revenue']

total_brand_revenue = data.groupby('Brand')['brand_revenue'].sum()
top_10_brands = total_brand_revenue.sort_values(ascending=False).head(10)
print(top_10_brands)
```

## Q 9.Compare performance regionwise

## a)	Draw a lineplot for the monthly Revenue of E-Commerce Company for each region separately.  (4 marks) 

## b)	Identify the best and the worst performing months for each region. (2 marks)

#### Note: Only those days with actual orders(Order_Date) placed are present in the dataset. Assuming there were no orders on other days. Also, show Month and Year on x-axis.

### Ans 9 a)

```python
data['Order_Date'] = pd.to_datetime(data['Order_Date'])
data['Month'] = data['Order_Date'].dt.to_period('M')

monthly_revenue = data.groupby(['Region', 'Month'])['Revenue'].sum().reset_index()

monthly_revenue['Month'] = monthly_revenue['Month'].astype(str)

plt.figure(figsize=(12, 6))

regions = data['Region'].unique()

for region in regions:
    region_data = monthly_revenue[monthly_revenue['Region'] == region]
    plt.plot(region_data['Month'], region_data['Revenue'], label=region)

plt.title('Monthly Revenue by Region')
plt.xlabel('Month')
plt.ylabel('Revenue')
plt.legend()
plt.grid(True)
plt.show()
```

### Ans 9 b)

```python
print("Best Month in east region: December 2020")
print("Worst Month in east region: July 2021")
```
